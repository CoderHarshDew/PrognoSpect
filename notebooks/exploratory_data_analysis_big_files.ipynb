# EDA For Big Files

## Import

```python
import os

import numpy as np
import pandas as pd
from pathlib import Path
from src.core.config import config_loader
import matplotlib.pyplot as plt
```

## Complete Analysis

```python
file_names = ['Friday-02-03-2018_TrafficForML_CICFlowMeter.csv', 'Friday-16-02-2018_TrafficForML_CICFlowMeter.csv', 'Friday-23-02-2018_TrafficForML_CICFlowMeter.csv', 'Thuesday-20-02-2018_TrafficForML_CICFlowMeter.csv', 'Thursday-01-03-2018_TrafficForML_CICFlowMeter.csv', 'Thursday-15-02-2018_TrafficForML_CICFlowMeter.csv', 'Thursday-22-02-2018_TrafficForML_CICFlowMeter.csv', 'Wednesday-14-02-2018_TrafficForML_CICFlowMeter.csv', 'Wednesday-21-02-2018_TrafficForML_CICFlowMeter.csv', 'Wednesday-28-02-2018_TrafficForML_CICFlowMeter.csv'
              ]

raw_dataset_dir = Path('../dataset/raw')
```

```python
current_dataset_file_idx = 3
```

```python
i = 1

for chunk in pd.read_csv(Path(raw_dataset_dir, file_names[current_dataset_file_idx]), low_memory=False, chunksize=1000000):
    header = chunk.columns

    header_rows = (chunk.astype(str) == header).all(axis=1)

    chunk = chunk.loc[~header_rows].copy()

    schema_cfg = config_loader('../config/preprocessing/validation_schema.yaml')

    non_numeric_cols = schema_cfg['non_numeric_col']
    numeric_cols = schema_cfg['numeric_col']

    cols_to_include = non_numeric_cols + numeric_cols
    cols_to_drop = chunk.columns.difference(cols_to_include)

    chunk.drop(columns=cols_to_drop, inplace=True)

    feature_cols = chunk.columns.intersection(numeric_cols)

    chunk[feature_cols] = chunk[feature_cols].apply(
        pd.to_numeric,
        errors="coerce"
    )

    chunk.info()

    chunk.replace([np.inf, -np.inf], np.nan, inplace=True)
    chunk.dropna(inplace=True)

    pd.set_option("display.float_format", "{:.2f}".format)
    chunk.max(numeric_only=True)
    print(chunk.describe())

    negative_counts = (chunk.select_dtypes(include="number") < 0).sum()
    print(negative_counts[negative_counts > 0])

    for col in chunk.columns:
        if chunk[col].dtype in ['int64', 'float64']:
            print(col, chunk[chunk[col] < 0][col].unique())

    print(chunk.columns)

    plt_path = Path('../images',  file_names[current_dataset_file_idx])
    plt_path.mkdir(parents=True, exist_ok=True)

    plt.scatter(chunk['Pkt Size Avg'], chunk['Pkt Len Mean'])
    plt.title("Pkt Len Mean Copy Check")
    plt.xlabel("Pkt Size Avg")
    plt.ylabel("Pkt Len Mean")
    plt.savefig(Path(plt_path,f'Pkt Len Mean Copy Check - {i}.png'))
    plt.show()

    for col in chunk.columns:
        if chunk[col].dtype in ['int64', 'float64'] and chunk[col].var() == 0:
            print(col, end=" ")
            print(chunk[col].unique())

    corr = chunk.corr(numeric_only=True)
    out_data_path = Path('../out/data', file_names[current_dataset_file_idx])
    out_data_path.mkdir(exist_ok=True, parents=True)
    corr.to_csv(Path(out_data_path, f'corr - {i}.csv'))
    print(corr)
    i += 1
```

From what can be seen here the findings  of this CSV are consistent with that of the  other CSVs.
